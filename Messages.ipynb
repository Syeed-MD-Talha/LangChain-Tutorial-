```python
api_key="Your gemini api key here"

from langchain.chat_models import init_chat_model
model=init_chat_model("gemini-2.0-flash",model_provider="google_genai",api_key=api_key)
```

## HumanMessage

```python
from langchain_core.messages import HumanMessage

response=model.invoke([HumanMessage(content="What is the capital of Bangladesh?")])
response.content
```

# SystemMessage & HumanMessage & AI message

```python
from langchain_core.messages import HumanMessage,SystemMessage
message=[
    SystemMessage(content="You are a singer and your name is Bob"),
    HumanMessage(content="Write a song on River")
]

response=model.invoke(message)  ## this is actually AI message what we call as response
response.content
```

# AIMessageChunk

```python
for chunk in model.stream([HumanMessage("what color is the blood?")]):
    print(chunk.content)
```

### Chunk Aggregating

```python
aggregation=[chunk.content for chunk in model.stream([HumanMessage("what color is the blood?")])]
ai_message = "".join(aggregation)
ai_message
```

## HumanMessage and AIMessage in another format

```python
response=model.invoke([
    {
        "role": "user",
        "content": "Hello, how are you?",
    },
    {
        "role": "assistant",
        "content": "I'm doing well, thank you for asking.",
    },
    {
        "role": "user",
        "content": "Can you tell me a joke?",
    }
])

print(response.content)
```
